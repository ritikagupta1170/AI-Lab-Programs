```python
import heapq

def manhattan(state, goal):
    distance = 0

    for i in range(3):
        for j in range(3):
            if state[i][j] != 0:
                for x in range(3):
                    for y in range(3):
                        if goal[x][y] == state[i][j]:
                            distance += abs(i - x) + abs(j - y)

    return distance


def get_neighbors(state):
    neighbors = []

    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                x, y = i, j

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dx, dy in moves:
        nx = x + dx
        ny = y + dy

        if 0 <= nx < 3 and 0 <= ny < 3:
            new_state = [row[:] for row in state]

            new_state[x][y], new_state[nx][ny] = new_state[nx][ny], new_state[x][y]

            neighbors.append(new_state)

    return neighbors


def astar(initial, goal):
    pq = []

    start = tuple(sum(initial, []))
    target = tuple(sum(goal, []))

    h = manhattan(initial, goal)

    heapq.heappush(pq, (h, 0, start))

    parent = {start: None}
    cost = {start: 0}

    while pq:
        f, g, current = heapq.heappop(pq)

        if current == target:
            path = []

            while current is not None:
                path.append(current)
                current = parent[current]

            return path[::-1]

        current_state = [
            list(current[0:3]),
            list(current[3:6]),
            list(current[6:9])
        ]

        for neighbor in get_neighbors(current_state):

            neighbor_tuple = tuple(sum(neighbor, []))

            new_cost = g + 1

            if neighbor_tuple not in cost or new_cost < cost[neighbor_tuple]:

                cost[neighbor_tuple] = new_cost

                h = manhattan(neighbor, goal)
                f = new_cost + h

                heapq.heappush(pq, (f, new_cost, neighbor_tuple))

                parent[neighbor_tuple] = current

    return None


initial = [
    [5, 4, 8],
    [6, 1, 2],
    [7, 3, 0]
]

goal = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]

path = astar(initial, goal)

if path:
    print("Solution Path:")

    for state in path:
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print()
else:
    print("No solution exists.")

```
