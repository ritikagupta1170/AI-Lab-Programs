```python
def unify(x, y, subs={}):
    if x == y:
        return subs

    if isinstance(x, str) and x.startswith("?"):
        if x in y:
            return None
        subs[x] = y
        return subs

    if isinstance(y, str) and y.startswith("?"):
        if y in x:
            return None
        subs[y] = x
        return subs

    if isinstance(x, tuple) and isinstance(y, tuple):
        if len(x) != len(y) or x[0] != y[0]:
            return None

        for a, b in zip(x[1:], y[1:]):
            subs = unify(a, b, subs)
            if subs is None:
                return None
        return subs

    return None


p1 = input("Enter first expression: ").split()
p2 = input("Enter second expression: ").split()

result = unify(tuple(p1), tuple(p2), {})

if result is None:
    print("Unification Failed")
else:
    print("Unification Successful")
    print("Substitution:", result)
```
