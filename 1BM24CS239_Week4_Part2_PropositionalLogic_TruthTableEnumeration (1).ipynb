```python
from itertools import product

def knowledge_base(P, Q, R):
    return ((not Q or P) and
            (not P or not Q) and
            (Q or R))

def entails(query, symbols):
    for values in product([True, False], repeat=len(symbols)):
        model = dict(zip(symbols, values))

        P = model['P']
        Q = model['Q']
        R = model['R']

        if knowledge_base(P, Q, R) and not query(P, Q, R):
            return False

    return True

print("P Q R | Q->P P->~Q Q|R | KB")

for P, Q, R in product([True, False], repeat=3):
    a = not Q or P
    b = not P or not Q
    c = Q or R
    kb = knowledge_base(P, Q, R)

    def val(x):
        return "T" if x else "F"

    print(val(P), val(Q), val(R), "|",
          val(a), val(b), val(c), "|", val(kb))

queries = [
    ("R", lambda P, Q, R: R),
    ("R -> P", lambda P, Q, R: not R or P),
    ("Q -> R", lambda P, Q, R: not Q or R)
]

print("\nEntailment Results:")

for name, query in queries:
    if entails(query, ['P', 'Q', 'R']):
        print("KB entails", name)
    else:
        print("KB does not entail", name)


```
