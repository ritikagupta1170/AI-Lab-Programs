```python

import random

# Calculate the number of attacking queen pairs
def calculate_cost(board, N):
    cost = 0

    for i in range(N):
        for j in range(i + 1, N):
            # Same row or same diagonal
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                cost += 1

    return cost


# Display the chessboard
def display_board(board, N):
    for row in range(N):
        for col in range(N):
            if board[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()


# Hill Climbing Algorithm
def hill_climbing():
    N = int(input("Enter the number of queens (N): "))

    # Place one queen in each column at a random row
    board = [random.randint(0, N - 1) for _ in range(N)]

    print("\nInitial Board:")
    display_board(board, N)

    while True:
        current_cost = calculate_cost(board, N)

        print("\nCurrent cost:", current_cost)

        # Solution found
        if current_cost == 0:
            print("\nSolution found!")
            display_board(board, N)
            return

        best_board = board[:]
        best_cost = current_cost

        # Generate neighboring states
        for col in range(N):
            for row in range(N):
                if row != board[col]:
                    neighbor = board[:]
                    neighbor[col] = row

                    cost = calculate_cost(neighbor, N)

                    if cost < best_cost:
                        best_cost = cost
                        best_board = neighbor

        # No better neighbor: local minimum
        if best_cost >= current_cost:
            print("\nStuck at a local minimum!")
            print("No better neighboring state is available.")
            print("Current board:")
            display_board(board, N)
            return

        # Move to the best neighboring state
        board = best_board


# Start the program
hill_climbing()
```
