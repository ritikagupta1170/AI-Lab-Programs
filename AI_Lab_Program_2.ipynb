```python
# Vacuum Cleaner Agent
# Classical AI Vacuum Cleaner Problem

class VacuumCleaner:
    def __init__(self, location, room_a, room_b):
        self.location = location
        self.rooms = {
            "A": room_a,
            "B": room_b
        }

    def display_state(self):
        print("\nCurrent State:")
        print(f"Vacuum Cleaner Location : Room {self.location}")
        print(f"Room A                 : {self.rooms['A']}")
        print(f"Room B                 : {self.rooms['B']}")

    def clean(self):
        """Clean the current room if it is dirty."""
        if self.rooms[self.location] == "Dirty":
            print(f"Action: SUCK -> Cleaning Room {self.location}")
            self.rooms[self.location] = "Clean"
        else:
            print(f"Room {self.location} is already clean.")

    def move_left(self):
        """Move from B to A."""
        if self.location == "B":
            print("Action: MOVE LEFT -> Room A")
            self.location = "A"
        else:
            print("Vacuum is already in Room A.")

    def move_right(self):
        """Move from A to B."""
        if self.location == "A":
            print("Action: MOVE RIGHT -> Room B")
            self.location = "B"
        else:
            print("Vacuum is already in Room B.")

    def is_goal_reached(self):
        """Check whether both rooms are clean."""
        return (
            self.rooms["A"] == "Clean"
            and self.rooms["B"] == "Clean"
        )

    def run(self):
        """Run the vacuum cleaner agent."""
        print("\n=== Vacuum Cleaner Agent ===")

        self.display_state()

        # Keep working until both rooms are clean
        while not self.is_goal_reached():

            # Clean current room
            self.clean()

            # Move to the other room
            if self.location == "A":
                self.move_right()
            else:
                self.move_left()

            self.display_state()

        print("\nGoal Reached!")
        print("Both Room A and Room B are CLEAN.")


# -------------------------------------------------
# Main Program
# -------------------------------------------------

# Initial state:
# Vacuum is in Room A
# Room A is Dirty
# Room B is Dirty

agent = VacuumCleaner(
    location="A",
    room_a="Dirty",
    room_b="Dirty"
)

agent.run()
```
